## manual tests for shape classes

```python
from shapes.rectangle import Rectangle
from shapes.circle import Circle
```

```python
circle1 = Circle(radius = 1, xpos = 0, ypos = 0) # unit circle
circle2 = Circle(radius = 1, xpos = 1, ypos = 1) 

rectangle = Rectangle(width=1, height=1, xpos=0, ypos=0)

print(circle1 == circle2) # True
print(circle2 == rectangle) # False

# translate, e.g. using a method or you could do other solution for translating
circle1.translate(5, 3) # moves its center 5 points in x and 3 points in y
try:
    circle1.translate("THREE", 5) # raise TypeError with an appropriate message
except TypeError as e:
    print(e)

circle3 = Circle(3) # a circle with center 0,0 with radius 3
print(circle3 >= circle1) # True

try:
    rectangle2 = Rectangle(width=3, height="5") # raise TypeError
except TypeError as e:
    print(e)

# use a for loop to add several circles and rectangles areas together
shape_list = (Circle(4), Circle(2), Rectangle(3,4), Rectangle(5,2))

tot_area = 0
for shape in shape_list:
    tot_area += shape.area
print(tot_area)
```

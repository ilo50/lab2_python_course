```python
from shapes.rectangle import Rectangle
from shapes.circle import Circle 
from shapes.cube import Cube
from shapes.sphere import Sphere

from plotters.rectangle_plotter import Rectangle_Plotter
from plotters.circle_plotter import Circle_Plotter
from plotters.cube_plotter import Cube_Plotter
from plotters.sphere_plotter import Sphere_Plotter

from plotters.plotter2D import Shape2dPlotter

from figures.circle_figure import Circle_Figure
from figures.rectangle_figure import Rectangle_Figure

import random
```

# 2D

## Rectangle plot test

```python
r1 = Rectangle(10, 20, -30, 0)

Rectangle_Plotter.draw(r1)
```

## Circle plot test

```python
ci1 = Circle(4, 0, 0)

Circle_Plotter.draw(ci1)
```

## 2D_plotter test


```python
c1 = Circle(4,3,1)
r1 = Rectangle(5,4,2,6)

Shape2dPlotter.plot(Rectangle_Figure(r1), Circle_Figure(c1))
```

## 2D_plotter with many figures

```python
circ_fig_list = []
rect_fig_list = []

for _ in range (10):

    circ_fig_list.append(Circle_Figure(Circle(random.randint(1,5), random.randint(-10,10), random.randint(-10,10))))
    rect_fig_list.append(Rectangle_Figure(Rectangle(random.randint(1,10), random.randint(1,10), random.randint(-10,10), random.randint(-10,10))))

fig_list = circ_fig_list + rect_fig_list

Shape2dPlotter.plot(*fig_list)
```
